```python
import pandas as pd
import numpy as np

file_path = "./Оптовые цены на мясо и птицу/WholesalePrices.xlsx"

df = pd.read_excel(
    file_path,
    header=None,
    dtype=str
)

print("Размер датасета:", df.shape)

print("\nПервые 10 строк:")
display(df.head(10))

print("\nТипы данных:")
print(df.dtypes)

print("\nКоличество полностью пустых строк:", df.isna().all(axis=1).sum())

print("\nКоличество полностью пустых столбцов:", df.isna().all(axis=0).sum())
```

```python
print("Первые значения первого столбца:")
display(df.iloc[:20, 0])

print("\nПоследние значения первого столбца:")
display(df.iloc[-20:, 0])

print("\nКоличество строк:")
print(len(df))

print("\nКоличество столбцов:")
print(len(df.columns))
```

```python
# Проверка размерности
print("Размер датасета:", df.shape)

# Проверка пустых строк
empty_rows = df.isna().all(axis=1)
print("Полностью пустых строк:", empty_rows.sum())

# Проверка пустых столбцов
empty_columns = df.isna().all(axis=0)
print("Полностью пустых столбцов:", empty_columns.sum())

# Проверка уникальных названий строк
print("\nКоличество уникальных названий:")
print(df.iloc[:, 1].nunique())

# Проверка дат в заголовках
print("\nНазвания столбцов:")
display(df.columns.tolist())

# Проверка количества NaN по столбцам
print("\nПропуски по столбцам:")
display(df.isna().sum())
```

```python
# Просмотр строк с числовыми значениями
numeric_data = df.iloc[:, 2:].apply(
    pd.to_numeric,
    errors="coerce"
)

has_prices = numeric_data.notna().any(axis=1)

print("Строк с ценовыми значениями:", has_prices.sum())

print("\nСтроки, содержащие цены:")
display(df.loc[has_prices].iloc[:, :8])
```

```python
# Первый столбец содержит наименования продуктов,
# остальные столбцы — значения цен
numeric_data = df.iloc[:, 1:].apply(
    pd.to_numeric,
    errors="coerce"
)

# Выбираем строки, в которых есть хотя бы одна цена
has_prices = numeric_data.notna().any(axis=1)

# Проверяем пропуски только в строках с ценами
price_data = numeric_data.loc[has_prices]

print("Количество строк с ценами:", len(price_data))

print("\nКоличество пропусков по каждому столбцу:")
display(price_data.isna().sum())

print("\nОбщее количество пропущенных цен:")
print(price_data.isna().sum().sum())

print("\nСтроки с пропущенными ценами:")
display(
    df.loc[
        has_prices & numeric_data.isna().any(axis=1)
    ]
)
```

```python
print("Минимальные цены:")
display(price_data.min())

print("\nМаксимальные цены:")
display(price_data.max())

print("\nКоличество нулевых значений:")
print((price_data == 0).sum().sum())

print("\nКоличество отрицательных значений:")
print((price_data < 0).sum().sum())
```

```python
# Создание рабочей копии
df_clean = df.copy()

# Количество строк до очистки
rows_before = len(df_clean)

# Удаление полностью пустых строк
df_clean = df_clean.dropna(how="all").reset_index(drop=True)

# Результат
print("Строк до очистки:", rows_before)
print("Строк после очистки:", len(df_clean))
print("Удалено пустых строк:", rows_before - len(df_clean))

display(df_clean.head(10))
```

```python
# Количество пропусков по каждому столбцу
print("Пропуски по столбцам:")
display(df_clean.isna().sum())

# Строки с пропущенными ценами
display(df_clean[df_clean.isna().any(axis=1)])
```

```python
# Выделяем строки, содержащие хотя бы одну цену
df_prices = df_clean[
    df_clean.iloc[:, 1:].notna().any(axis=1)
].copy()

# Сбрасываем индексы
df_prices = df_prices.reset_index(drop=True)

print("Количество ценовых записей:", len(df_prices))

display(df_prices)
```

```python
# Сохраняем названия периодов
periods = df_prices.iloc[0, 1:].tolist()

# Удаляем строку Wholesale prices
df_prices = df_prices.iloc[1:].reset_index(drop=True)

# Переименовываем столбцы
df_prices.columns = ["Product"] + periods

# Добавляем категории
category_map = {
    "Choice 1-3, 600-900 lbs": "Beef",
    "Select 1-3, 600-900 lbs": "Beef",
    "Boneless, 90 percent, fresh": "Beef",
    "Imported boneless, 90 percent, frozen": "Beef",

    "Pork cutout composite": "Pork",
    'Loins, 1/4", trimmed vacuum-packed': "Pork",
    "Bellies, 12-14 lb, skin on": "Pork",
    "Hams, 23-27 lb, trimmed selected": "Pork",
    "Trimmings, 72 percent combo": "Pork",

    "National composite": "Broilers",
    "Breast, boneless": "Broilers",
    "Breast, with ribs": "Broilers",
    "Legs, whole": "Broilers",
    "Leg quarters": "Broilers",

    "Hens, 8-16 lbs": "Turkeys",
    "Toms, 16-24 lbs": "Turkeys",
    "Breast, 4-8 lbs": "Turkeys",
    "Drumsticks": "Turkeys",
    "Wings, full cut": "Turkeys",

    "Combined regional": "Eggs",
    "New York": "Eggs"
}

df_prices["Category"] = df_prices["Product"].map(category_map)

# Проверяем, все ли товары получили категорию
print("Товары без категории:")
display(df_prices[df_prices["Category"].isna()])

# Добавляем единицы измерения
unit_map = {
    "Beef": "dollars/cwt",
    "Pork": "dollars/cwt",
    "Broilers": "cents/lb",
    "Turkeys": "cents/lb",
    "Eggs": "cents/dozen"
}

df_prices["Unit"] = df_prices["Category"].map(unit_map)

# Переставляем столбцы
df_prices = df_prices[
    ["Product", "Category", "Unit"] + periods
]

display(df_prices)
```

```python
# Удаляем старые столбцы, если они уже существуют
df_prices = df_prices.drop(
    columns=["Category", "Unit"],
    errors="ignore"
)

# Назначаем категории по порядку строк
categories = (
    ["Beef"] * 4 +
    ["Pork"] * 5 +
    ["Broilers"] * 5 +
    ["Turkeys"] * 5 +
    ["Eggs"] * 2
)

# Назначаем единицы измерения
units = (
    ["dollars/cwt"] * 9 +
    ["cents/lb"] * 10 +
    ["cents/dozen"] * 2
)

# Проверяем количество строк
assert len(df_prices) == len(categories) == len(units)

df_prices.insert(1, "Category", categories)
df_prices.insert(2, "Unit", units)

display(df_prices)
```

```python
print("Размер таблицы:", df_prices.shape)
print("Количество товаров:", df_prices["Product"].nunique())
print("Количество категорий:", df_prices["Category"].nunique())

print("\nКоличество пропусков:")
display(df_prices.isna().sum())

print("\nКоличество товаров по категориям:")
display(df_prices["Category"].value_counts())

print("\nПроверка единиц измерения:")
display(df_prices.groupby("Category")["Unit"].unique())
```

```python
import os

output_dir = "processed_datasets/USDA_ERS_Wholesale_Prices"
os.makedirs(output_dir, exist_ok=True)

df_prices.to_csv(
    os.path.join(output_dir, "USDA_ERS_Wholesale_Prices_Clean.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("Количество записей:", len(df_prices))
print("Количество столбцов:", len(df_prices.columns))
print("Файл сохранён:", output_dir)
```
