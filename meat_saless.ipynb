```python
import pandas as pd
import numpy as np

file_path = "./Оптовые цены на мясо и птицу/WholesalePrices.xlsx"

df = pd.read_excel(
    file_path,
    header=None,
    dtype=str
)

print("Размер датасета:", df.shape)

print("\nПервые 10 строк:")
display(df.head(10))

print("\nТипы данных:")
print(df.dtypes)

print("\nКоличество полностью пустых строк:", df.isna().all(axis=1).sum())

print("\nКоличество полностью пустых столбцов:", df.isna().all(axis=0).sum())
```

```python
print("Первые значения первого столбца:")
display(df.iloc[:20, 0])

print("\nПоследние значения первого столбца:")
display(df.iloc[-20:, 0])

print("\nКоличество строк:")
print(len(df))

print("\nКоличество столбцов:")
print(len(df.columns))
```

```python
# Проверка размерности
print("Размер датасета:", df.shape)

# Проверка пустых строк
empty_rows = df.isna().all(axis=1)
print("Полностью пустых строк:", empty_rows.sum())

# Проверка пустых столбцов
empty_columns = df.isna().all(axis=0)
print("Полностью пустых столбцов:", empty_columns.sum())

# Проверка уникальных названий строк
print("\nКоличество уникальных названий:")
print(df.iloc[:, 1].nunique())

# Проверка дат в заголовках
print("\nНазвания столбцов:")
display(df.columns.tolist())

# Проверка количества NaN по столбцам
print("\nПропуски по столбцам:")
display(df.isna().sum())
```

```python
# Просмотр строк с числовыми значениями
numeric_data = df.iloc[:, 2:].apply(
    pd.to_numeric,
    errors="coerce"
)

has_prices = numeric_data.notna().any(axis=1)

print("Строк с ценовыми значениями:", has_prices.sum())

print("\nСтроки, содержащие цены:")
display(df.loc[has_prices].iloc[:, :8])
```

```python
# Первый столбец содержит наименования продуктов,
# остальные столбцы — значения цен
numeric_data = df.iloc[:, 1:].apply(
    pd.to_numeric,
    errors="coerce"
)

# Выбираем строки, в которых есть хотя бы одна цена
has_prices = numeric_data.notna().any(axis=1)

# Проверяем пропуски только в строках с ценами
price_data = numeric_data.loc[has_prices]

print("Количество строк с ценами:", len(price_data))

print("\nКоличество пропусков по каждому столбцу:")
display(price_data.isna().sum())

print("\nОбщее количество пропущенных цен:")
print(price_data.isna().sum().sum())

print("\nСтроки с пропущенными ценами:")
display(
    df.loc[
        has_prices & numeric_data.isna().any(axis=1)
    ]
)
```

```python
print("Минимальные цены:")
display(price_data.min())

print("\nМаксимальные цены:")
display(price_data.max())

print("\nКоличество нулевых значений:")
print((price_data == 0).sum().sum())

print("\nКоличество отрицательных значений:")
print((price_data < 0).sum().sum())
```
