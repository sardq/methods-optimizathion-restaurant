```python
import pandas as pd

file_path = "./Сроки хранения/FoodKeeper-Data.xls"  # укажи путь к своему файлу

xls = pd.ExcelFile(file_path)

print("Количество листов:", len(xls.sheet_names))
print("Названия листов:", xls.sheet_names)

sheets = {}

for sheet_name in xls.sheet_names:
    df = pd.read_excel(file_path, sheet_name=sheet_name)
    sheets[sheet_name] = df

    print(f"\n{'=' * 60}")
    print(f"Лист: {sheet_name}")
    print("Размер таблицы:", df.shape)
    print("\nТипы данных:")
    print(df.dtypes)

    print("\nПропущенные значения:")
    missing = pd.DataFrame({
        "Пропуски": df.isna().sum(),
        "Доля пропусков, %": (df.isna().mean() * 100).round(2)
    })
    display(missing[missing["Пропуски"] > 0])

    print("Полные дубликаты:", df.duplicated().sum())

    print("\nПервые 5 строк:")
    display(df.head())
```

```python
product = sheets["Product"].copy()

# Основные показатели сроков хранения
storage_columns = [
    col for col in product.columns
    if col.endswith("_Min")
    or col.endswith("_Max")
    or col.endswith("_Metric")
]

# Полнота данных по срокам хранения
storage_quality = pd.DataFrame({
    "Заполнено": product[storage_columns].notna().sum(),
    "Пропущено": product[storage_columns].isna().sum(),
    "Доля пропусков, %": (
        product[storage_columns].isna().mean() * 100
    ).round(2)
})

display(storage_quality)

# Проверка числовых сроков хранения
numeric_columns = [
    col for col in product.columns
    if col.endswith("_Min") or col.endswith("_Max")
]

for col in numeric_columns:
    values = pd.to_numeric(product[col], errors="coerce")

    print(f"\n{col}")
    print("Количество значений:", values.notna().sum())
    print("Минимум:", values.min())
    print("Максимум:", values.max())
    print("Нулевые значения:", (values == 0).sum())
    print("Отрицательные значения:", (values < 0).sum())
```

```python
# Уникальность ID
for name in ["Category", "Product", "CookingTips", "CookingMethods"]:
    df = sheets[name]

    print(
        f"{name}: всего ID = {df['ID'].notna().sum()}, "
        f"уникальных ID = {df['ID'].nunique()}, "
        f"дубликатов ID = {df['ID'].duplicated().sum()}"
    )

# Проверка ссылок Product -> Category
category_ids = set(sheets["Category"]["ID"])
product_category_ids = set(sheets["Product"]["Category_ID"].dropna())

print(
    "\nНесуществующие категории:",
    product_category_ids - category_ids
)

# Проверка ссылок CookingTips и CookingMethods -> Product
product_ids = set(product["ID"])

for name in ["CookingTips", "CookingMethods"]:
    referenced_ids = set(sheets[name]["Product_ID"].dropna())

    print(
        f"{name}: ссылки на отсутствующие продукты:",
        referenced_ids - product_ids
    )
```

```python
# Проверка минимальных и максимальных сроков
pairs = [
    ("Pantry_Min", "Pantry_Max", "Pantry_Metric"),
    ("DOP_Pantry_Min", "DOP_Pantry_Max", "DOP_Pantry_Metric"),
    ("Pantry_After_Opening_Min", "Pantry_After_Opening_Max",
     "Pantry_After_Opening_Metric"),
    ("Refrigerate_Min", "Refrigerate_Max", "Refrigerate_Metric"),
    ("DOP_Refrigerate_Min", "DOP_Refrigerate_Max",
     "DOP_Refrigerate_Metric"),
    ("Refrigerate_After_Opening_Min", "Refrigerate_After_Opening_Max",
     "Refrigerate_After_Opening_Metric"),
    ("Refrigerate_After_Thawing_Min", "Refrigerate_After_Thawing_Max",
     "Refrigerate_After_Thawing_Metric"),
    ("Freeze_Min", "Freeze_Max", "Freeze_Metric"),
    ("DOP_Freeze_Min", "DOP_Freeze_Max", "DOP_Freeze_Metric")
]

for min_col, max_col, metric_col in pairs:
    check = product[
        product[min_col].notna() & product[max_col].notna()
    ]

    invalid = check[check[min_col] > check[max_col]]

    different_metrics = check[
        check[metric_col].isna()
    ]

    print(f"\n{min_col} — {max_col}")
    print("Проверено диапазонов:", len(check))
    print("Минимум больше максимума:", len(invalid))
    print("Отсутствует единица измерения:", len(different_metrics))

    if len(invalid) > 0:
        display(invalid[["Name", min_col, max_col, metric_col]])
```
