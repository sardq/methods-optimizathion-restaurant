```python
import fitz
import pandas as pd
import os

files = {
    "Мясо и мясные продукты": "./Коэффициент выхода продуктов/USDA_FBG_Section1_MeatsAndMeatAlternatesYieldTable.pdf",
    "Овощи": "./Коэффициент выхода продуктов/USDA_FBG_Section2_VegetablesYieldTable.pdf",
    "Фрукты": "./Коэффициент выхода продуктов/USDA_FBG_Section3_FruitsYieldTable.pdf",
    "Прочие продукты": "./Коэффициент выхода продуктов/USDA_FBG_Section6_OtherFoodsYieldTable.pdf"
}

pdf_data = {}

for name, path in files.items():
    if not os.path.exists(path):
        print(f"Файл не найден: {path}")
        continue

    doc = fitz.open(path)
    pages = [page.get_text() for page in doc]
    full_text = "\n".join(pages)

    pdf_data[name] = {
        "pages": pages,
        "text": full_text,
        "page_count": len(doc)
    }

    print(f"\n{name}")
    print(f"Количество страниц: {len(doc)}")
    print(f"Количество символов: {len(full_text)}")
    print(f"Количество строк: {len(full_text.splitlines())}")
    print(f"Пустых страниц: {sum(not page.strip() for page in pages)}")
```

```python
# Старые заголовки мясного раздела
meat_old_headings = df_fbg_headings[
    df_fbg_headings["Section"] == "Мясо и мясные продукты"
][["Section", "Page", "LineNumber", "Product"]].copy()

# Новые заголовки смешанного регистра
meat_new_headings = meat_product_candidates[
    ["Section", "Page", "LineNumber", "RawText"]
].copy()

meat_new_headings = meat_new_headings.rename(
    columns={"RawText": "Product"}
)

# Объединяем заголовки
meat_headings = pd.concat(
    [meat_old_headings, meat_new_headings],
    ignore_index=True
).drop_duplicates(
    subset=["Section", "Page", "LineNumber"]
)

meat_headings = meat_headings.sort_values(
    ["Page", "LineNumber"]
)

# Привязываем коэффициенты к продуктам
meat_yield_records = []
current_product = None

for _, row in meat.sort_values(
    ["Page", "LineNumber"]
).iterrows():

    heading = meat_headings[
        (meat_headings["Page"] == row["Page"]) &
        (meat_headings["LineNumber"] == row["LineNumber"])
    ]

    if not heading.empty:
        heading_product = heading.iloc[0]["Product"]

        if not re.search(
            r"\(continued\)\s*$",
            str(heading_product),
            re.IGNORECASE
        ):
            current_product = heading_product

    if ap_pattern.search(row["RawText"]):
        meat_yield_records.append({
            "Section": row["Section"],
            "Product": current_product,
            "Page": row["Page"],
            "LineNumber": row["LineNumber"],
            "APDescription": row["RawText"]
        })

df_fbg_meat_yields = pd.DataFrame(meat_yield_records)

display(
    df_fbg_meat_yields.groupby("Product")
    .agg(
        CoefficientRows=("APDescription", "size"),
        UniqueDescriptions=("APDescription", "nunique")
    )
    .reset_index()
)
```

```python
meat_sorted = meat.sort_values(
    ["Page", "LineNumber"]
).reset_index(drop=True)

meat_ap_records = []

incomplete_pattern = re.compile(
    r"(,|;|\(|\b(?:cooked|heated|drained|boned|lean|trimmed|sliced|with|without|about|approximately))\s*$",
    re.IGNORECASE
)

for i, row in meat_sorted.iterrows():

    if not ap_pattern.search(row["RawText"]):
        continue

    description = row["RawText"].strip()

    for j in range(i + 1, len(meat_sorted)):

        next_row = meat_sorted.iloc[j]
        next_text = next_row["RawText"].strip()

        if next_row["Page"] != row["Page"]:
            break

        if (
            ap_pattern.search(next_text)
            or next_text.upper() == "POUND"
            or re.match(r"^\d+(?:\.\d+)?$", next_text)
            or re.match(r"^\d+\s*(?:oz|lb|cup)\b", next_text, re.IGNORECASE)
        ):
            break

        has_unclosed_parenthesis = (
            description.count("(") > description.count(")")
        )

        if not incomplete_pattern.search(description) and not has_unclosed_parenthesis:
            break

        if next_text and next_text.lstrip()[0].islower():
            description += " " + next_text
        else:
            break

    meat_ap_records.append({
        "Page": row["Page"],
        "LineNumber": row["LineNumber"],
        "APDescription": description
    })

df_fbg_meat_ap = pd.DataFrame(meat_ap_records)

display(df_fbg_meat_ap.head(30))
```

```python
product_start_pattern = re.compile(
    r"^(?:Beans|Bean Products|Beef|Cheese|Chicken|Eggs|Game|"
    r"Lamb|Lentils|Mutton|Pork|Seafood|Seeds|Tempeh|Tofu|"
    r"Turkey|Veal|Yogurt|Tree Nuts|Other Nuts|Peas|Pea Soup),",
    re.IGNORECASE
)

product_titles = []

for page in meat_sorted["Page"].unique():

    page_df = meat_sorted[
        meat_sorted["Page"] == page
    ].reset_index(drop=True)

    for i, row in page_df.iterrows():

        text = row["RawText"].strip()

        if not product_start_pattern.match(text):
            continue

        product_name = text

        # Присоединяем продолжение названия,
        # если строка заканчивается на запятую или "or".
        if i + 1 < len(page_df):
            next_text = page_df.iloc[i + 1]["RawText"].strip()

            if (
                product_name.endswith(",")
                or product_name.lower().endswith("or")
            ):
                product_name += " " + next_text

        product_titles.append({
            "Page": page,
            "LineNumber": row["LineNumber"],
            "Product": product_name
        })

df_meat_product_titles = pd.DataFrame(product_titles)

display(
    df_meat_product_titles[
        df_meat_product_titles["Page"].isin([2, 7])
    ].to_string(index=False)
)
```
