```python
import fitz
import pandas as pd
import os

files = {
    "Мясо и мясные продукты": "./Коэффициент выхода продуктов/USDA_FBG_Section1_MeatsAndMeatAlternatesYieldTable.pdf",
    "Овощи": "./Коэффициент выхода продуктов/USDA_FBG_Section2_VegetablesYieldTable.pdf",
    "Фрукты": "./Коэффициент выхода продуктов/USDA_FBG_Section3_FruitsYieldTable.pdf",
    "Прочие продукты": "./Коэффициент выхода продуктов/USDA_FBG_Section6_OtherFoodsYieldTable.pdf"
}

pdf_data = {}

for name, path in files.items():
    if not os.path.exists(path):
        print(f"Файл не найден: {path}")
        continue

    doc = fitz.open(path)
    pages = [page.get_text() for page in doc]
    full_text = "\n".join(pages)

    pdf_data[name] = {
        "pages": pages,
        "text": full_text,
        "page_count": len(doc)
    }

    print(f"\n{name}")
    print(f"Количество страниц: {len(doc)}")
    print(f"Количество символов: {len(full_text)}")
    print(f"Количество строк: {len(full_text.splitlines())}")
    print(f"Пустых страниц: {sum(not page.strip() for page in pages)}")
```

```python
summary = pd.DataFrame([
    {
        "Раздел": name,
        "Количество страниц": data["page_count"],
        "Количество символов": len(data["text"]),
        "Количество строк": len(data["text"].splitlines()),
        "Пустых страниц": sum(
            not page.strip() for page in data["pages"]
        )
    }
    for name, data in pdf_data.items()
])

display(summary)
```

```python
import re
import pandas as pd

for name, data in pdf_data.items():
    numeric_values = []
    empty_lines = 0

    for page in data["pages"]:
        for line in page.splitlines():
            line = line.strip()

            if not line:
                empty_lines += 1
                continue

            # Числовые значения, вынесенные в отдельные строки
            if re.fullmatch(r"-?\d+(?:\.\d+)?", line):
                numeric_values.append(float(line))

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Количество числовых значений:", len(numeric_values))
    print("Количество пустых строк:", empty_lines)

    if numeric_values:
        print("Минимальное значение:", min(numeric_values))
        print("Максимальное значение:", max(numeric_values))
        print("Нулевых значений:", sum(x == 0 for x in numeric_values))
        print("Отрицательных значений:", sum(x < 0 for x in numeric_values))
```

```python
missing_patterns = [
    r"\bN/A\b",
    r"\bNA\b",
    r"\bNOT AVAILABLE\b",
    r"\bNOT REPORTED\b",
    r"\bUNKNOWN\b"
]

for name, data in pdf_data.items():
    print(f"\n{name}")

    for pattern in missing_patterns:
        matches = re.findall(
            pattern,
            data["text"],
            flags=re.IGNORECASE
        )

        if matches:
            print(f"{pattern}: {len(matches)}")
```

```python
patterns = [
    "AP =",
    "EP",
    "ready-to-serve",
    "cooked",
    "drained",
    "peeled"
]

for name, data in pdf_data.items():
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    for pattern in patterns:
        matches = [
            line.strip()
            for line in data["text"].splitlines()
            if pattern.lower() in line.lower()
        ]

        print(f"\n{pattern}: {len(matches)} упоминаний")

        for line in matches[:5]:
            print(line)
```
