```python
import pandas as pd

file_path = "./Пищевые потери/Product-level_FLWdatabase_May2026.xlsx"

# Загружаем лист без автоматического определения заголовков
raw = pd.read_excel(
    file_path,
    sheet_name="Database",
    header=None
)

# Первая строка — названия столбцов
headers = raw.iloc[0].fillna("").astype(str).str.strip()

# Вторая строка — типы данных
data_types = raw.iloc[1]

# Начиная с третьей строки идут реальные наблюдения
db = raw.iloc[2:].copy()
db.columns = headers
db = db.reset_index(drop=True)

# Пустые строки преобразуем в пропуски
db = db.replace(r"^\s*$", pd.NA, regex=True)

print("Размер основной таблицы:", db.shape)

print("\nНазвания столбцов и указанные типы данных:")
for column, dtype in zip(db.columns, data_types):
    print(f"{column}: {dtype}")

# Проверка пропусков
missing = pd.DataFrame({
    "Пропуски": db.isna().sum(),
    "Доля пропусков, %": (db.isna().mean() * 100).round(2)
})

print("\nПропущенные значения:")
display(missing[missing["Пропуски"] > 0].sort_values(
    "Доля пропусков, %", ascending=False
))

# Проверка полных дубликатов
print("\nКоличество полных дубликатов:", db.duplicated().sum())

# Проверка первых строк
print("\nПервые 5 записей:")
display(db.head())

# Проверка количества уникальных значений
print("\nКоличество уникальных значений:")
display(db.nunique().sort_values(ascending=False))
```

```python
headers = [
    "Food group leaving primary production (FoodEx2)",
    "Product leaving primary production (FoodEx2)",
    "Product entering supply chain stage (FoodEx2)",
    "Product leaving supply chain stage (FoodEx2)",
    "Product leaving primary production (CPA 2.2)",
    "Product entering supply chain stage (CPA 2.2)",
    "Product leaving supply chain stage (CPA 2.2)",
    "Product specification based on study",
    "Country of data collection",
    "Year(s) of data collection",
    "Year of publication",
    "Supply chain stage (broad)",
    "Supply chain stage (specific)",
    "NACE code",
    "FLW Quantity 1",
    "Unit 1",
    "Edible/Inedible 1",
    "FLW Quantity 2",
    "Unit 2",
    "Edible/Inedible 2",
    "Variance",
    "Disposal or other use",
    "Feed, donations and others",
    "Data collection method",
    "Number of observations",
    "Sample description",
    "Uncertainty about data",
    "Further detailed description",
    "Source of the data"
]

db.columns = headers

print("Количество столбцов:", len(db.columns))
display(db.head())
```

```python
# Количество наблюдений по этапам цепочки поставок
print("Распределение по этапам цепочки поставок:")
display(
    db["Supply chain stage (broad)"]
    .value_counts(dropna=False)
    .to_frame("Количество наблюдений")
)

# Подробное распределение по категориям этапов
print("\nПодробные этапы:")
display(
    db["Supply chain stage (specific)"]
    .value_counts(dropna=False)
    .to_frame("Количество наблюдений")
)

# Распределение по странам
print("\nГеографическое распределение:")
display(
    db["Country of data collection"]
    .value_counts(dropna=False)
    .to_frame("Количество наблюдений")
)

# Распределение по годам
print("\nГоды сбора данных:")
display(
    db["Year(s) of data collection"]
    .value_counts(dropna=False)
    .to_frame("Количество наблюдений")
)
```

```python
# Выбираем записи, связанные с ресторанами и общественным питанием
restaurant_data = db[
    db["Supply chain stage (broad)"]
    .astype("string")
    .str.contains("restaurant|food service", case=False, na=False)
].copy()

print("Количество записей:", len(restaurant_data))

# Выводим наиболее важные для нашей задачи поля
columns_to_check = [
    "Product specification based on study",
    "Country of data collection",
    "Year(s) of data collection",
    "Supply chain stage (broad)",
    "Supply chain stage (specific)",
    "FLW Quantity 1",
    "Unit 1",
    "Edible/Inedible 1",
    "FLW Quantity 2",
    "Unit 2",
    "Edible/Inedible 2",
    "Variance",
    "Data collection method",
    "Number of observations",
    "Uncertainty about data",
    "Further detailed description",
    "Source of the data"
]

display(restaurant_data[columns_to_check])
```

```python
# Проверка количества и полноты данных в ресторанной выборке
print("Количество записей:", len(restaurant_data))

check_columns = [
    "Product specification based on study",
    "Country of data collection",
    "Year(s) of data collection",
    "Supply chain stage (broad)",
    "Supply chain stage (specific)",
    "FLW Quantity 1",
    "Unit 1",
    "Edible/Inedible 1",
    "Data collection method",
    "Number of observations",
    "Uncertainty about data",
    "Source of the data"
]

quality = pd.DataFrame({
    "Заполнено": restaurant_data[check_columns].notna().sum(),
    "Пропущено": restaurant_data[check_columns].isna().sum(),
    "Доля пропусков, %": (
        restaurant_data[check_columns].isna().mean() * 100
    ).round(2)
})

display(quality)

# Проверка используемых единиц измерения
print("Единицы измерения:")
display(restaurant_data["Unit 1"].value_counts(dropna=False))

# Проверка способов сбора данных
print("Методы сбора данных:")
display(restaurant_data["Data collection method"].value_counts(dropna=False))
```

```python
# Только рестораны и предприятия общественного питания
restaurants_only = db[
    db["Supply chain stage (broad)"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("restaurants and food services")
].copy()

# Рестораны и домохозяйства
restaurants_households = db[
    db["Supply chain stage (broad)"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("restaurants & households")
].copy()

print("Только рестораны:", len(restaurants_only))
print("Рестораны и домохозяйства:", len(restaurants_households))
```

```python
missing_stats = pd.DataFrame({
    "MissingCount": restaurants_only.isna().sum(),
    "MissingPercent": (
        restaurants_only.isna().sum() / len(restaurants_only) * 100
    ).round(2)
})

missing_stats = missing_stats[
    missing_stats["MissingCount"] > 0
].sort_values("MissingPercent", ascending=False)

print("Всего записей:", len(restaurants_only))
display(missing_stats)
```
